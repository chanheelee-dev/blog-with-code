```python
from collections import Counter, deque, namedtuple, ChainMap
```

```python
li = ['a', 'a', 'b', 'b', 'b', 'c', 'c', 'c', 'c']
c = Counter(li)

print(c["a"])
print(c.most_common(2))
print(list(c.elements()))
```

```python
dq = deque(li)
dq.appendleft('x')
dq.append('y')
print(dq)
print(dq.pop())
print(dq.popleft())
print(dq)

print(dq.rotate(1))
print(dq)

```

```python
Point = namedtuple('Point', 'x, y')
pt = Point(1, -4)
print(pt)
print(pt.x, pt.y)

# pt.x = 3  # can't set attribute
pt = pt._replace(x=3)
print(pt)
```

```python
dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 3, 'c': 4}

cm = ChainMap(dict1, dict2)
print(cm)
print(cm.maps)
print(cm["a"])
print(cm["b"])  # 순서 상 앞의 map에서 찾는 것이 우선
```
